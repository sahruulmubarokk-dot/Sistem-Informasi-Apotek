```python
n = int(input("Jumlah variabel/persamaan (n): "))

A = []
b = []
for i in range(n):
    baris = list(map(float, input(f"Koefisien persamaan {i+1} (pisah spasi): ").split()))
    A.append(baris)
    b.append(float(input(f"Konstanta persamaan {i+1}: ")))

print("\nData tersimpan.")
```

Sel 1 – Input data
Sel ini meminta data SPL dari pengguna.

n adalah jumlah persamaan sekaligus jumlah variabel (SPL harus berbentuk persegi agar solusinya tunggal).
Perulangan for membaca satu persamaan per putaran. input(...).split() memecah angka yang dipisah spasi, lalu map(float, ...) mengubahnya jadi bilangan desimal.
Koefisien satu persamaan disimpan sebagai satu baris di list A, dan konstantanya disimpan di list b.

```python
def tampil(M, judul=""):
    if judul:
        print(judul)
    for baris in M:
        print("  ".join(f"{v:9.4f}" for v in baris))
    print()
```

Sel 2 – Fungsi bantu tampil
Sel ini hanya membuat fungsi untuk mencetak matriks dengan rapi, tanpa menghitung apa pun.

f"{v:9.4f}" mencetak tiap angka selebar 9 karakter dengan 4 angka di belakang koma, jadi kolomnya sejajar.
Fungsi ini dipakai di Sel 3 dan Sel 4 untuk menampilkan matriks tiap tahap, sehingga proses eliminasinya terlihat langkah demi langkah.

```python
M = [list(map(float, A[i])) + [float(b[i])] for i in range(n)]
tampil(M, "Matriks augmented awal:")
```

Sel 3 – Matriks augmented [A | b]
Sel ini menggabungkan A dan b menjadi satu matriks M dengan b sebagai kolom terakhir, sehingga operasi baris cukup dilakukan pada satu matriks. Matriks 3 × 3 menjadi 3 × 4. float(...) memastikan semua nilai desimal supaya pembagiannya akurat. Hasil awalnya ditampilkan dengan tampil.

```python
singular = False

for k in range(n):
    # Cari pivot: baris dengan |nilai| terbesar pada kolom k
    p = max(range(k, n), key=lambda i: abs(M[i][k]))

    # Jika pivot = 0 maka tidak ada solusi tunggal
    if abs(M[p][k]) < 1e-12:
        singular = True
        print("Pivot = 0 pada kolom", k + 1, "-> SPL tidak punya solusi tunggal")
        break

    # Tukar baris k dengan baris pivot
    M[k], M[p] = M[p], M[k]

    # Nolkan semua elemen di bawah pivot
    for i in range(k + 1, n):
        faktor = M[i][k] / M[k][k]
        for j in range(k, n + 1):
            M[i][j] -= faktor * M[k][j]

    tampil(M, f"Setelah eliminasi kolom {k+1}:")
```

Sel 4 – Eliminasi maju
Sel ini mengubah M menjadi bentuk segitiga atas, yaitu semua elemen di bawah diagonal menjadi 0. Variabel k adalah kolom pivot yang sedang diproses.

Cari pivot. max(..., key=abs) mencari baris dari k ke bawah dengan nilai mutlak terbesar di kolom k. Ini pivot parsial, yang membuat hasil lebih stabil dan menghindari pembagian dengan nol.
Cek singular. Jika pivot mendekati 0 (< 1e-12), singular diisi True dan proses berhenti dengan break. Artinya SPL tidak punya solusi tunggal.
Tukar baris. M[k], M[p] = M[p], M[k] menukar baris k dengan baris pivot.
Nolkan elemen di bawah pivot. Untuk tiap baris i di bawah k, hitung faktor = M[i][k] / M[k][k], lalu kurangi baris i dengan faktor × baris k. Operasi ini disebut operasi baris elementer (OBE).
tampil(...) mencetak matriks setelah tiap kolom selesai, untuk menunjukkan prosesnya.

```python
x = [0.0] * n

if not singular:
    for i in range(n - 1, -1, -1):
        jumlah = sum(M[i][j] * x[j] for j in range(i + 1, n))
        x[i] = (M[i][n] - jumlah) / M[i][i]
else:
    print("Substitusi mundur dilewati karena matriks singular.")
```

Sel 5 – Substitusi mundur
Sel ini menghitung nilai variabel dari persamaan paling bawah ke atas, setelah matriks berbentuk segitiga atas.

x = [0.0] * n menyiapkan tempat penyimpan hasil.
Perulangan range(n-1, -1, -1) berjalan dari baris terakhir ke baris pertama.
Baris terakhir hanya punya satu variabel, jadi langsung bisa dihitung. Baris di atasnya memakai nilai x yang sudah diketahui. jumlah adalah total koefisien × x yang sudah ketahuan, lalu x[i] = (M[i][n] - jumlah) / M[i][i] menyelesaikan sisanya.
Jika singular bernilai True, sel ini hanya mencetak pesan dan tidak menghitung.

```python
if not singular:
    print("Solusi SPL:")
    for i in range(n):
        print(f"x{i+1} = {x[i]:.4f}")
```

Sel 6 – Tampilkan solusi
Sel ini mencetak x1 ... xn dengan 4 angka di belakang koma. Jika matriks singular, tidak ada yang dicetak karena pesannya sudah muncul di Sel 4 dan 5.

```python
if not singular:
    for i in range(n):
        hasil = sum(A[i][j] * x[j] for j in range(n))
        print(f"Persamaan {i+1}: ruas kiri = {hasil:.4f}, b = {b[i]}")
```

Sel 7 – Verifikasi
Sel ini mengecek hasil dengan memasukkan solusi x ke persamaan awal, memakai A dan b asli yang tidak ikut berubah selama eliminasi. Ruas kiri (Σ A[i][j] × x[j]) harus sama dengan b[i]. Selisih sangat kecil, misalnya 7.9999 vs 8, wajar karena pembulatan desimal.
